## Data Preprocessing
The Wisconsin Diagnostic Breast Cancer (WDBC) dataset was obtained from the UCI Machine Learning Repository using the ucimlrepo Python package (id=17). It contains 569 instances and 30 continuous features, extracted from digitized images of fine needle aspirate (FNA) of breast masses. Each instance is labeled as either malignant or benign.
The following preprocessing steps were applied:

1. **Class labels recoded:** the original diagnosis labels (M for malignant, B for benign) were recoded to integers: M = 1, B = 2.
2. **Z-normalization:** the 30 continuous features were z-normalized (zero mean, unit variance). This was necessary because the features vary significantly in scale;for example, area values range in the hundreds while smoothness values are near zero. Without normalization, large-scale features would dominate the Euclidean distance calculation in the 1-NN classifier.

```python
# pip install ucimlrepo
```

```python
from ucimlrepo import fetch_ucirepo
import numpy as np
```

```python
breast_cancer = fetch_ucirepo(id=17)
X = breast_cancer.data.features.values.astype(float)
y = breast_cancer.data.targets.values.flatten()
```

```python
y = np.where(y == 'M', 1, 2)
```

```python
X = (X - X.mean(axis=0)) / X.std(axis=0)
```

```python
data_clean = np.column_stack([y, X])
np.savetxt("wdbc_clean.txt", data_clean, fmt="%.7e")
```
